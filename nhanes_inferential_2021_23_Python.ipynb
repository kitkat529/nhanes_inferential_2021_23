```python
# Question 1: Is there an association between marital status and education level? Chi-square

import pandas as pd
from scipy.stats import chi2_contingency

df=pd.read_csv("nhanes_adults_clean.csv")

df["Marital_Status"].value_counts()
df["Education_level"].value_counts()

contingency_table = pd.crosstab(df["Marital_Status"], df["Education_level"])
print(contingency_table)

chi2, p, _, _ = chi2_contingency(contingency_table)
print(f"Chi2 value: {chi2}")
print(f"P-value: {p}")

## Answer: Yes, there is an association. P-value is much less than .05 and Chi2 value is large which
## further supports a strong association between the two variables "Marital_status" and "Education_level"
```

```python
# Question 2: Is there a difference in the mean sedentary behavior time between those who are married
# and those who are not married? T-test

import pandas as pd
from scipy.stats import ttest_ind

df=pd.read_csv("nhanes_adults_clean.csv")

married = df[df["Marital_Status"] == "Married"]["Minutes_Sedentary_Behavior"]
not_married = df[df["Marital_Status"] == "Not married"]["Minutes_Sedentary_Behavior"]

t_stat, p_val = ttest_ind(married, not_married, equal_var=False)
print(f"T-statistic: {t_stat}")
print(f"P-value: {p_val}")

# Answer: Yes, there is a statistically significant difference. Married people on average have less
# sedentary behavior time than those that are not married. The large T-statistic suggests the means
# are far apart from each other, the P-value less than .05 is statistically significant
```

```python
# Question 3: How do age and marital status affect systolic blood pressure? 2-Way ANOVA

import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd

df=pd.read_csv("nhanes_adults_clean.csv")

# Drop rows with missing values in the relevant columns for ANOVA and Tukey
df_anova = df.dropna(subset=["Systolic_BP", "Age_group", "Marital_Status"]).copy()

model = ols("Systolic_BP ~ C(Age_group) * C(Marital_Status)", data=df_anova).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)

# perform tukey test to see which age groups are different
tukey = pairwise_tukeyhsd(endog=df_anova["Systolic_BP"], groups=df_anova["Age_group"], alpha=0.05)
print(tukey)

# Answer: Age and marital status each independently have a statistically significan impact on systolic
# blood pressure, but the interaction of the two variables does not have a statistically significant
# impact as shown by the p value results PR(>F). The Tukey test results show that as age groups get older, the
# diffrence in the mean systolic BP also increases
```

```python
# Question 4: Is there a correlation between self reported weight and minutes of sedentary behavior? Regression

import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt

df=pd.read_csv("nhanes_adults_clean.csv")

# Missing data should have been removed during cleaning, but just in case
df = df.dropna(subset=["Self_Reported_Weight_lbs", "Minutes_Sedentary_Behavior"])

X = df['Self_Reported_Weight_lbs']
y = df['Minutes_Sedentary_Behavior']


# Add a constant to the independent variable (required for the statsmodels regression model)
X = sm.add_constant(X)

# Fit the regression model
model = sm.OLS(y, X).fit()

# Print the summary of the regression
print(model.summary())

plt.scatter(df['Self_Reported_Weight_lbs'], df['Minutes_Sedentary_Behavior'], label='Data Points')
plt.plot(df['Self_Reported_Weight_lbs'], model.predict(X), color='blue', label='Regression Line')
plt.xlabel('Self Reported Weight (lbs)')
plt.ylabel('Minutes of Sedentary Behavior')
plt.title('Relationship between Self Reported Weight and Minutes of Sedentary Behavior')
plt.legend()
plt.show()

# Answer: Yes, there is a correlation. As self reported weight increases, minutes of sedentary behavior
# also increases
```

```python
# Question 5: Is there an association between diabetes and weak or failing kidneys? Chi-square

import pandas as pd
from scipy.stats import chi2_contingency

df=pd.read_csv("nhanes_adults_clean.csv")

df["Diabetes"].value_counts()
df["Weak/Failing_Kidneys"].value_counts()

contingency_table = pd.crosstab(df["Diabetes"], df["Weak/Failing_Kidneys"])
print(contingency_table)

chi2, p, _, _ = chi2_contingency(contingency_table)
print(f"Chi2 value: {chi2}")
print(f"P-value: {p}")

# Answer: Yes, the results suggest that having diabetes is associated with a higher likelihood of
# having weak or failing kidneys
```
